```python
DB_PATH = "../data/chroma_db"

client = chromadb.PersistentClient(path=DB_PATH)

collection = client.get_collection("policy_chunks")

print("Collection:", collection.name)
print("Total Chunks:", collection.count())
```

```python
def retrieve(query, top_k=5):

    query_embedding = model.encode(query).tolist()

    results = collection.query(

        query_embeddings=[query_embedding],

        n_results=top_k
    )

    return results
```

```python
def show_results(query):

    results = retrieve(query)

    print("=" * 80)
    print("QUERY:")
    print(query)
    print("=" * 80)

    docs = results["documents"][0]
    ids = results["ids"][0]
    distances = results["distances"][0]
    metadatas = results["metadatas"][0]

    for i in range(len(docs)):

        similarity = 1 - distances[i]

        metadata = metadatas[i]

        print(f"\nRank {i+1}")

        print(f"Chunk ID : {ids[i]}")

        print(f"Similarity : {similarity:.4f}")

        print(f"File : {metadata.get('file_name','N/A')}")

        print(f"Page : {metadata.get('page_number','N/A')}")

        print("\nChunk Text")

        print("-"*60)

        print(docs[i][:600])

        print("-"*60)
```

```python
queries = [

    "asset inventory",

    "access control",

    "role based access control",

    "backup policy",

    "incident response",

    "risk assessment",

    "password policy",

    "audit logs",

    "encryption",

    "business continuity"

]
```

```python
results = retrieve("asset inventory")

print(results.keys())

print("\nMetadata:")
print(results["metadatas"])

print("\nDocuments:")
print(results["documents"])

print("\nDistances:")
print(results["distances"])
```

```python
for q in queries:

    show_results(q)

    print("\n\n")
```

```python
records = []

for query in queries:

    results = retrieve(query, top_k=1)

    metadata = results["metadatas"][0][0]

    distance = results["distances"][0][0]

    similarity = 1-distance

    records.append({

        "Query": query,

        "Retrieved File": metadata["file_name"],

        "Page": metadata["page_number"],

        "Similarity": round(similarity,3)

    })

df = pd.DataFrame(records)

df
```

```python
print("Average Similarity Score")

print(df["Similarity"].mean())
```

```python
high = len(df[df["Similarity"]>=0.80])

medium = len(df[(df["Similarity"]>=0.60) & (df["Similarity"]<0.80)])

low = len(df[df["Similarity"]<0.60])

print(f"High Similarity : {high}")

print(f"Medium Similarity : {medium}")

print(f"Low Similarity : {low}")
```
